```python
# Diabetes Model Comparison in Google Colab
# Enhanced version to match paper's visualization style with MLP added

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

results = [
    {"Model": "CFA", "Dataset": "PID", "Accuracy": 0.7619, "Kappa": 0.4598, "MAE": 0.2381, "Time": 32.40},
    {"Model": "CFA", "Dataset": "HFD", "Accuracy": 0.7950, "Kappa": 0.5295, "MAE": 0.2050, "Time": 38.12},
    {"Model": "GA", "Dataset": "PID", "Accuracy": 0.7749, "Kappa": 0.4969, "MAE": 0.2251, "Time": 6.07},
    {"Model": "GA", "Dataset": "HFD", "Accuracy": 0.9650, "Kappa": 0.9237, "MAE": 0.0350, "Time": 6.94},
    {"Model": "MLP", "Dataset": "PID", "Accuracy": 0.7403, "Kappa": 0.4160, "MAE": 0.2597, "Time": 20.8919},
    {"Model": "MLP", "Dataset": "HFD", "Accuracy": 0.8167, "Kappa": 0.5926, "MAE": 0.1833, "Time": 33.4594},
]
results_df = pd.DataFrame(results)


sns.set_style('whitegrid')
sns.set()

plt.rcParams.update({
    'font.size': 12,
    'axes.titlesize': 14,
    'axes.labelsize': 12,
    'xtick.labelsize': 11,
    'ytick.labelsize': 11,
    'legend.fontsize': 11
})

# --- Figure 1: Accuracy Comparison ---
plt.figure(figsize=(10, 6))
bar_width = 0.25
index = np.arange(2)  # PID and HFD datasets

cfa_acc = results_df[results_df['Model'] == 'CFA']['Accuracy'].values
ga_acc = results_df[results_df['Model'] == 'GA']['Accuracy'].values
mlp_acc = results_df[results_df['Model'] == 'MLP']['Accuracy'].values

bars1 = plt.bar(index, cfa_acc, bar_width, label='CFA', color='#1f77b4')
bars2 = plt.bar(index + bar_width, ga_acc, bar_width, label='GA', color='#ff7f0e')
bars3 = plt.bar(index + 2*bar_width, mlp_acc, bar_width, label='MLP', color='#2ca02c')

plt.xlabel('Dataset')
plt.ylabel('Accuracy')
plt.title('Accuracy Comparison Across Models and Datasets')
plt.xticks(index + bar_width, ('PID', 'HFD'))
plt.ylim(0.7, 1.0)
plt.legend(loc='upper left')

for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                 f'{height:.2f}',
                 ha='center', va='bottom')

plt.tight_layout()
plt.show()

# --- Figure 2: Kappa Comparison ---
plt.figure(figsize=(10, 6))

cfa_kappa = results_df[results_df['Model'] == 'CFA']['Kappa'].values
ga_kappa = results_df[results_df['Model'] == 'GA']['Kappa'].values
mlp_kappa = results_df[results_df['Model'] == 'MLP']['Kappa'].values

bars1 = plt.bar(index, cfa_kappa, bar_width, label='CFA', color='#1f77b4')
bars2 = plt.bar(index + bar_width, ga_kappa, bar_width, label='GA', color='#ff7f0e')
bars3 = plt.bar(index + 2*bar_width, mlp_kappa, bar_width, label='MLP', color='#2ca02c')

plt.xlabel('Dataset')
plt.ylabel('Kappa Statistic')
plt.title('Kappa Statistic Comparison Across Models and Datasets')
plt.xticks(index + bar_width, ('PID', 'HFD'))
plt.ylim(0.3, 1.0)
plt.legend(loc='upper left')

for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                 f'{height:.2f}',
                 ha='center', va='bottom')

plt.tight_layout()
plt.show()

# --- Figure 3: MAE Comparison ---
plt.figure(figsize=(10, 6))

cfa_mae = results_df[results_df['Model'] == 'CFA']['MAE'].values
ga_mae = results_df[results_df['Model'] == 'GA']['MAE'].values
mlp_mae = results_df[results_df['Model'] == 'MLP']['MAE'].values

bars1 = plt.bar(index, cfa_mae, bar_width, label='CFA', color='#1f77b4')
bars2 = plt.bar(index + bar_width, ga_mae, bar_width, label='GA', color='#ff7f0e')
bars3 = plt.bar(index + 2*bar_width, mlp_mae, bar_width, label='MLP', color='#2ca02c')

plt.xlabel('Dataset')
plt.ylabel('Mean Absolute Error (MAE)')
plt.title('MAE Comparison Across Models and Datasets')
plt.xticks(index + bar_width, ('PID', 'HFD'))
plt.ylim(0, 0.3)
plt.legend(loc='upper right')

for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                 f'{height:.2f}',
                 ha='center', va='bottom')

plt.tight_layout()
plt.show()

# --- Figure 4: Execution Time Comparison ---
plt.figure(figsize=(10, 6))

cfa_time = results_df[results_df['Model'] == 'CFA']['Time'].values
ga_time = results_df[results_df['Model'] == 'GA']['Time'].values
mlp_time = results_df[results_df['Model'] == 'MLP']['Time'].values

bars1 = plt.bar(index, cfa_time, bar_width, label='CFA', color='#1f77b4')
bars2 = plt.bar(index + bar_width, ga_time, bar_width, label='GA', color='#ff7f0e')
bars3 = plt.bar(index + 2*bar_width, mlp_time, bar_width, label='MLP', color='#2ca02c')

plt.xlabel('Dataset')
plt.ylabel('Execution Time (seconds)')
plt.title('Execution Time Comparison Across Models and Datasets')
plt.xticks(index + bar_width, ('PID', 'HFD'))
plt.legend(loc='upper left')

for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height,
                 f'{height:.1f}',
                 ha='center', va='bottom')

plt.tight_layout()
plt.show()

# --- Performance Summary Table ---
print("\nPerformance Summary Table:")
summary_table = results_df.pivot(index="Model", columns="Dataset", values=["Accuracy", "Kappa", "MAE", "Time"])
print(summary_table.to_markdown(floatfmt=".3f"))
```
